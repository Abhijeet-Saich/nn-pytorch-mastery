```python
import torch
import torch.nn.functional as F
```

```python
data = open('names.txt', 'r').read().splitlines()    #reading file and splitting it
print(len(data), data[0:10])
```

```python
# SETTING UP MAPPING DICTIONARIES
alphabets = sorted(set("".join(data)))

stoi = {ch : i+1 for i,ch in enumerate(alphabets)}
stoi['.'] = 0
itos = {i+1 : x for i,x in enumerate(stoi)}
```

```python
# CREATING DATASET
words = data

X = []
Y = []

for w in words:
    context = [0,0,0]
    for ch in w + '.':
        X.append(context)
        chi = stoi[ch]
        Y.append(chi)
        context = context[1:] + [chi]
        
X = torch.tensor(X)
Y = torch.tensor(Y)
X.shape, Y.shape
```

```python
#CREATING TRAINING DATASET

n = int(0.9*len(X))
Xtr = X[:n]
Xdev = X[n:]

Ytr = Y[:n]
Ydev = Y[n:]
n, Xtr.shape, Ytr.shape, Xdev.shape, Ydev.shape 
```

```python
# CREATING NEURAL NETWORK MATRICES

C = torch.randn(27,10)

#LAYER no.1, 200 neurons with 30 weights
w1 = torch.randn(200, 30)
b1 = torch.randn(200)

#Layer no.2, 27 neurons with 200 w's each
w2 = torch.randn(27, 200)
b2 = torch.randn(27)

parameters = [w1, b1, w2, b2]
for p in parameters:
    p.requires_grad = True

w1.shape, b1.shape, w2.shape, b2.shape
```

```python
# FORWARD PASS FUNCTION

def forwardPass(X, Y):
     inp = C[X].view(-1,30) #single input represeneted by 30 values (same as w's of single neuron)
     h1 = torch.tanh(inp @ w1.T + b1)  # operation performed after addin bias
    
     logits = h1 @ w2.T + b2
     return F.cross_entropy(logits, Y)
```

```python
# To SAMPLE RANDOM ALPHABET

def randomCh(xi):
    with torch.no_grad():
        inp = C[xi].view(-1,30)           # single input represeneted by 30 values (same as w's of single neuron)
        h1 = torch.tanh(inp @ w1.T + b1)  # operation performed after addin bias
        logits = h1 @ w2.T + b2
        probs = torch.softmax(logits,1)
        return torch.multinomial(probs, num_samples=1).item()
    
```

```python
gen_names_raw = []  # list to store generated names



while(len(gen_names_raw) < 40):
    context = [0] * 3
    name = []

    #finding index for start of name
    while(True):
        ind = randomCh(context)
        if ind == 0 and len(name) >= 3 :
            gen_names_raw.append("".join(name))
            break
        elif(ind == 0):
            pass
        else:
            ch = itos[ind]
            name.append(ch)
            context = context[1:] + [ind]
        

print(gen_names_raw)
```

```python
train_loss = forwardPass(Xtr, Ytr)
dev_loss = forwardPass(Xdev, Ydev)
# test_loss = forwardPass(Xte, Yte)

print("train:", train_loss.item())
print("dev:", dev_loss.item())
# print("test:", test_loss.item())
```

```python
#GRADIENT DECENT LOOP

epochs = 100000 + 1
l_r = 0.01    #learning rate

for i in range(epochs):
    if i > 30000:
        l_r = 0.0001    #learning rate
    samples = torch.randint(0, Xtr.shape[0], (32,))    #sample 32 datapoints from trainging set
    
    imp = Xtr[samples] #single input represeneted by 30 values (same as w's of single neuron)
    out = Ytr[samples]
    
    loss = forwardPass(imp, out)
    loss.backward()  # backpropagation
    
    #updating parameters using gradient decent
    with torch.no_grad():
        for p in parameters:
            p -= l_r * p.grad
    
    for p in parameters:
        p.grad = None   # resetting gradients 
    
    if (i % 5000 == 0):
        print('Loss After iteration ',i ,' :', loss.item())
```

```python
gen_names = []  # list to store generated names



while(len(gen_names) < 40):
    context = [0] * 3
    name = []

    #finding index for start of name
    while(True):
        ind = randomCh(context)
        if ind == 0 and len(name) >= 3 :
            gen_names.append("".join(name))
            break
        elif(ind == 0):
            pass
        else:
            ch = itos[ind]
            name.append(ch)
            context = context[1:] + [ind]
        

print(gen_names_raw)
print(gen_names)
```
