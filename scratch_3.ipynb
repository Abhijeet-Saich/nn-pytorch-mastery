```python
import torch
import matplotlib.pyplot as plt
```

```python
x = torch.linspace(-10, 10 , 1000)
print(x.shape,x)
```

```python
y = torch.tanh(x)
print(y.shape)
count = 0
for k in y:
    if k > 0.97:
        count += 1

print(count/1000 * 100)
```

```python
plt.plot(x, y)
plt.grid()
```

```python
dy = 1 - y**2
plt.plot(x, dy)
```

```python

x = torch.tensor(10.0, requires_grad=True)

w = torch.tensor(15.0, requires_grad=True)
b = torch.tensor(20.0, requires_grad=True)

h = x * w + b
h.retain_grad()
print(h.item())

logit = torch.tanh(h)
logit.retain_grad()
print('logit', logit.item())

loss = logit**2
print('loss', loss.item())
```

```python
loss.backward()
logit.grad.item(),h.grad.item(), w.grad.item(), b.grad.item()
```

```python
x = torch.tensor(1.0)

w = torch.tensor(0.1, requires_grad=True)
b = torch.tensor(-0.1, requires_grad=True)

z = w * x + b
y = torch.tanh(z)

target = torch.tensor(1.0)
loss = (y - target)**2

loss.backward()

print("z =", z.item())
print("y =", y.item())
print("loss =", loss.item())
print("dloss/dw =", w.grad.item())
print("dloss/db =", b.grad.item())
```

```python
x = torch.tensor([10., 11, 17, 18, 21])
mean = x.mean()

variance = ((x-mean)**2).mean()

std = variance.sqrt()

normy = (x - mean)/std
normy, variance, normy.mean(), normy.std(unbiased=False)
```
