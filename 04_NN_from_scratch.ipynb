```python
import torch
```

```python
x = torch.tensor(2.0)
w = torch.tensor(3.0, requires_grad=True)
b = torch.tensor(2.0, requires_grad=True)

target = torch.tensor(10.0, requires_grad=True)

learning_rate = 0.01

for _ in range(40):


    #forward pass
    pred = w * x + b

    #calculating loss
    loss = (target - pred)**2
    print(loss.data)

    # doing backpropagation
    loss.backward()

    # updating the w parameter
    with torch.no_grad():
        w -= learning_rate * w.grad
        b -= learning_rate * b.grad

    #setting up gradient zero 
    w.grad.zero_()
        
  
```

```python
import torch

# Dataset
x = torch.tensor([1., 2., 3., 4., 5.])
y = torch.tensor([5., 8., 11., 14., 17.])

# Parameters
w = torch.randn(1, requires_grad=True)
b = torch.randn(1, requires_grad=True)

learning_rate = 0.01

for step in range(1000):

    # ----------------
    # Forward pass
    # ----------------
    prediction = x * w + b

    # ----------------
    # Loss
    # ----------------
    loss = ((prediction - y) ** 2).mean()

    # ----------------
    # Backpropagation
    # ----------------
    loss.backward()

    # ----------------
    # Gradient descent
    # ----------------
    with torch.no_grad():
        w -= learning_rate * w.grad
        b -= learning_rate * b.grad

    # ----------------
    # Clear gradients
    # ----------------
    w.grad.zero_()
    b.grad.zero_()

    if step % 100 == 0:
        print(
            step,
            "loss =", loss.item(),
            "w =", w.item(),
            "b =", b.item()
        )
```

```python
x = torch.tensor([1., 2., 3., 4., 5.])
y = torch.tensor([5., 8., 11., 14., 17.])

w = torch.tensor([3.0], requires_grad=True)
b = torch.tensor([2.0], requires_grad=True)

prediction = x * w + b

loss = ((prediction - y) ** 2).mean()

loss.backward()

print("PyTorch dw:", w.grad)
print("PyTorch db:", b.grad)
```
