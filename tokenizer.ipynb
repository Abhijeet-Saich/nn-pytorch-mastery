```python
#Character level tokenizer

class CharTokenizer:

    def __init__(self):
        self.stoi = None
        self.itos = None

    def train(self,text):
        chars = sorted(set(text))
        self.stoi = {ch : i for i, ch in enumerate(chars)}
        self.itos = {i : ch for ch, i in self.stoi.items()}

    def encode(self,text):
        return [self.stoi[ch] for ch in text]
        
    def decode(self,ids):
        return ''.join([self.itos[i] for i in ids])
```

```python
text = """Complete the W3Schools Python course, strengthen your knowledge, and earn a certificate you can add to your CV, portfolio, and LinkedIn profile."""

corpus = CharTokenizer()
corpus.train(text)

print(corpus.stoi)
print('\n')
print(corpus.itos)

print(corpus.encode(text))
print(corpus.decode(corpus.encode(text)))
print(corpus.decode(corpus.encode(text)) == text)

```

```python
def get_stats(ids):
    counts = {}
    for x, y in zip(ids,ids[1:]):
        counts[(x,y)] = counts.get((x,y), 0) + 1

    return counts



def merge(ids, pair, idx):
    arr = []
    i = 0
    while(i < len(ids)):
        if (i < len(ids) -1) and (ids[i], ids[i+1]) == pair:
            arr.append(idx)
            i += 2
        else:
            arr.append(ids[i])
            i += 1
            
        
    return arr    
```

```python
def train(text, vocab_size=None):
    byte_ids = list(text.encode("utf-8"))
    new_id = 256
    merges = {}
    vocab = {i: bytes([i]) for i in range(256)}

    for i in range(vocab_size - 256   #difference is the no of new vocab added
        counts = get_stats(byte_ids)
        if not counts:
            break
        a, b = max(counts, key=counts.get)
        vocab[new_id] = vocab[a] + vocab[b]
        
        byte_ids = merge(byte_ids, (a,b), new_id)

        # adding th entries
        merges[(a,b)] = new_id
        #increase the id by 1 for new merging
        new_id += 1
    
    return merges, vocab
```

```python
merges, vocab = train("aaavvvv", 268)
print(merges, '\n' ,vocab)
```

```python
def decode(ids, vocab):
    raw = b"".join(vocab[id] for id in ids)
    return raw.decode("utf-8", errors="replace")
```

```python
def encode(text, merges):
    
    ids = list(text.encode("utf-8"))
    while len(ids) >= 2:
        stats = get_stats(ids)
        pair = min(stats, key=lambda p: merges.get(p, float("inf")))
        if pair not in merges:                 
            break
        ids = merge(ids, pair, merges[pair])
    return ids    
```

```python
merges, vocab = train("aaavvvv", 300)
for s in ["aaavvvv", "aaa", "vva", "héllo 😀"]:
    ids = encode(s, merges)
    print(s, ids, decode(ids, vocab) == s)
```
