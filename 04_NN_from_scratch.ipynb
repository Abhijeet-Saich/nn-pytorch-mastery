```python
x = torch.tensor(2.0)
w = torch.tensor(3.0, requires_grad=True)
b = torch.tensor(2.0, requires_grad=True)

target = torch.tensor(10.0, requires_grad=True)

learning_rate = 0.01

for _ in range(40):


    #forward pass
    pred = w * x + b

    #calculating loss
    loss = (target - pred)**2
    print(loss.data)

    # doing backpropagation
    loss.backward()

    # updating the w parameter
    with torch.no_grad():
        w -= learning_rate * w.grad
        b -= learning_rate * b.grad

    #setting up gradient zero 
    w.grad.zero_()
        
  
```

```python
import torch

# Dataset
x = torch.tensor([1., 2., 3., 4., 5.])
y = torch.tensor([5., 8., 11., 14., 17.])

# Parameters
w = torch.randn(1, requires_grad=True)
b = torch.randn(1, requires_grad=True)

learning_rate = 0.01

for step in range(1000):

    # ----------------
    # Forward pass
    # ----------------
    prediction = x * w + b

    # ----------------
    # Loss
    # ----------------
    loss = ((prediction - y) ** 2).mean()

    # ----------------
    # Backpropagation
    # ----------------
    loss.backward()

    # ----------------
    # Gradient descent
    # ----------------
    with torch.no_grad():
        w -= learning_rate * w.grad
        b -= learning_rate * b.grad

    # ----------------
    # Clear gradients
    # ----------------
    w.grad.zero_()
    b.grad.zero_()

    if step % 100 == 0:
        print(
            step,
            "loss =", loss.item(),
            "w =", w.item(),
            "b =", b.item()
        )
```

```python
import torch

x = torch.tensor([1.0, 2.0, 3.0])

W1 = torch.randn(4, 3, requires_grad=True)
b1 = torch.zeros(4, requires_grad=True)

W2 = torch.randn(2, 4, requires_grad=True)
b2 = torch.zeros(2, requires_grad=True)

# Forward
z1 = W1 @ x + b1
a1 = torch.relu(z1)

z2 = W2 @ a1 + b2
a2 = torch.relu(z2)

target = torch.tensor([1.0, 0.0])

loss = ((a2 - target) ** 2).mean()

print("loss:", loss.item())

loss.backward()

print("W1 gradient:")
print(W1.grad)

print("b1 gradient:")
print(b1.grad)

print("W2 gradient:")
print(W2.grad)

print("b2 gradient:")
print(b2.grad)
```

```python
import torch

# -----------------------------
# Dataset
# -----------------------------

X = torch.tensor([
    [0., 0.],
    [0., 1.],
    [1., 0.],
    [1., 1.]
])

y = torch.tensor([
    [0.],
    [1.],
    [1.],
    [0.]
])


# -----------------------------
# Parameters
# -----------------------------

W1 = torch.randn(2, 2, requires_grad=True)
b1 = torch.zeros(2, requires_grad=True)

W2 = torch.randn(1, 2, requires_grad=True)
b2 = torch.zeros(1, requires_grad=True)


learning_rate = 0.1


# -----------------------------
# Training
# -----------------------------

for step in range(10000):

    # Forward pass

    z1 = X @ W1.T + b1
    a1 = torch.relu(z1)

    z2 = a1 @ W2.T + b2
    y_hat = torch.sigmoid(z2)


    # Binary cross entropy

    eps = 1e-7

    loss = -(
        y * torch.log(y_hat + eps)
        + (1 - y) * torch.log(1 - y_hat + eps)
    ).mean()


    # Backward pass

    loss.backward()


    # Gradient descent

    with torch.no_grad():

        W1 -= learning_rate * W1.grad
        b1 -= learning_rate * b1.grad

        W2 -= learning_rate * W2.grad
        b2 -= learning_rate * b2.grad


    # Clear gradients

    W1.grad.zero_()
    b1.grad.zero_()
    W2.grad.zero_()
    b2.grad.zero_()


    if step % 1000 == 0:
        print(
            f"step={step}, "
            f"loss={loss.item():.4f}"
        )


# -----------------------------
# Final prediction
# -----------------------------

with torch.no_grad():

    z1 = X @ W1.T + b1
    a1 = torch.relu(z1)

    z2 = a1 @ W2.T + b2
    y_hat = torch.sigmoid(z2)

    print("\nPredictions:")
    print(y_hat)
```
