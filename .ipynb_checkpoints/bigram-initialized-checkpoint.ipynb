# Bigram, initialised: does the neural bigram reach the count-based optimum?

A bigram model has an **exact answer**: the best possible bigram is just the count table
normalised into probabilities. That makes it a perfect experiment for checking a neural model against ground truth:

1. Build the 65×65 count matrix from `train_data`, add 1 to every cell (smoothing), normalise each row,
   and compute the cross-entropy on the validation set. That number is the **floor** for any bigram model.
2. Retrain the neural bigram (lr = 1e-2, 20,000 steps) from **three different initialisations** and check
   that its validation loss approaches the floor and its `Q` row approaches the true `P(U | Q) ≈ 0.94`.

```python
import time
import torch
import torch.nn as nn
from torch.nn import functional as F
import matplotlib.pyplot as plt

torch.manual_seed(1337)

with open('shakespere.txt', 'r', encoding='utf-8') as f:
    text = f.read()

chars = sorted(list(set(text)))
vocab_size = len(chars)
stoi = {ch: i for i, ch in enumerate(chars)}
itos = {i: ch for i, ch in enumerate(chars)}
encode = lambda s: [stoi[c] for c in s]
decode = lambda l: ''.join([itos[i] for i in l])

data = torch.tensor(encode(text), dtype=torch.long)
n = int(0.9 * len(data))
train_data, val_data = data[:n], data[n:]
print("vocab_size:", vocab_size, "| train:", len(train_data), "| val:", len(val_data))
```

## 1. The count-based bigram: the exact optimum

`counts[i, j]` = how many times character `j` follows character `i`.
Normalising each row gives `P[i, j] = P(next = j | current = i)`.

**Why +1 smoothing?** Some pairs appear in val but never in train. Without smoothing, P = 0 for those,
so −ln P = ∞ and the val loss would be infinite. Adding 1 to every cell gives every pair a small non-zero probability.

```python
def count_matrix(d):
    # counts[i, j] += 1 for every adjacent pair (d[k], d[k+1])
    c = torch.zeros((vocab_size, vocab_size))
    c.index_put_((d[:-1], d[1:]), torch.ones(len(d) - 1), accumulate=True)
    return c

train_counts = count_matrix(train_data)
val_counts   = count_matrix(val_data)

# Pairs in val that never occur in train: these are why smoothing is needed
unseen = ((train_counts == 0) & (val_counts > 0)).sum().item()
print(f"pairs that appear in val but never in train: {unseen}")

def ce_from_counts(logP, counts):
    # Exact cross-entropy over a whole split, using its pair counts:
    # mean over all pairs of -log P[i, j]  ==  sum(counts * -logP) / total pairs
    return (-(counts * logP).sum() / counts.sum()).item()

P_counts = (train_counts + 1) / (train_counts + 1).sum(dim=1, keepdim=True)   # +1 smoothing, row-normalise
logP_counts = P_counts.log()

floor_train = ce_from_counts(logP_counts, train_counts)
floor_val   = ce_from_counts(logP_counts, val_counts)
print(f"count-based bigram  train loss: {floor_train:.4f}   val loss (THE FLOOR): {floor_val:.4f}")
print(f"uniform guessing    ln(65) = {torch.log(torch.tensor(65.)).item():.4f}")
```

## 2. The neural bigram and three ways to initialise it

The model is the same one-table model as in `my_llm_1.ipynb`. The only thing that changes is the **starting value of the table W**:

| init | W at step 0 | what softmax(W) predicts at step 0 |
|---|---|---|
| `random` | N(0, 1), PyTorch's default for `nn.Embedding` | random, sometimes confidently wrong |
| `zeros` | all 0 | exactly uniform: 1/65 for every next char |
| `counts` | log P from the count table | exactly the count-based optimum |

The `counts` init works because if `W[i] = log P[i]`, then `softmax(W[i]) = P[i] / sum(P[i]) = P[i]`.
The neural model can **represent** the count model exactly; the question is whether gradient descent **finds** it.

Losses here are computed **exactly** over the full split with `ce_from_counts`, not from sampled batches,
so they compare directly with the floor.

```python
class BigramLanguageModel(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.token_embedding_table = nn.Embedding(vocab_size, vocab_size)

    def forward(self, idx, targets=None):
        logits = self.token_embedding_table(idx)            # (B, T, C)
        if targets is None:
            return logits, None
        B, T, C = logits.shape
        loss = F.cross_entropy(logits.view(B*T, C), targets.view(B*T))
        return logits, loss

def make_model(init):
    model = BigramLanguageModel(vocab_size)                 # default init: N(0, 1)
    W = model.token_embedding_table.weight
    with torch.no_grad():                                   # editing weights is not part of the graph
        if init == 'zeros':
            W.zero_()
        elif init == 'counts':
            W.copy_(logP_counts)
    return model

def exact_losses(model):
    logP = F.log_softmax(model.token_embedding_table.weight.detach(), dim=-1)   # (65, 65)
    return ce_from_counts(logP, train_counts), ce_from_counts(logP, val_counts)

batch_size, block_size = 32, 8
def get_batch(split):
    d = train_data if split == 'train' else val_data
    ix = torch.randint(len(d) - block_size, (batch_size,))
    x = torch.stack([d[i:i+block_size] for i in ix])
    y = torch.stack([d[i+1:i+block_size+1] for i in ix])
    return x, y

for init in ['random', 'zeros', 'counts']:
    tr, va = exact_losses(make_model(init))
    print(f"{init:>7} init  step-0 loss  train {tr:.4f}  val {va:.4f}")
```

## 3. Train each one: lr = 1e-2, 20,000 steps

Same seed for each run, so all three see the same batches. Only the starting weights differ.

```python
def train(init, lr=1e-2, max_iters=20_000, eval_interval=500):
    torch.manual_seed(1337)
    model = make_model(init)
    W_start = model.token_embedding_table.weight.detach().clone()
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr)
    hist = {'step': [], 'train': [], 'val': []}
    t0 = time.time()
    for step in range(max_iters + 1):
        if step % eval_interval == 0:
            tr, va = exact_losses(model)
            hist['step'].append(step); hist['train'].append(tr); hist['val'].append(va)
            if step % 5000 == 0:
                print(f"[{init:>6}] step {step:>6}: train {tr:.4f}  val {va:.4f}")
        if step == max_iters:
            break
        xb, yb = get_batch('train')
        _, loss = model(xb, yb)
        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        optimizer.step()
    print(f"[{init:>6}] done in {time.time() - t0:.1f}s\n")
    return model, W_start, hist

runs = {init: train(init) for init in ['random', 'zeros', 'counts']}
```

## 4. Validation loss vs the floor

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.5))
colors = {'random': 'tab:red', 'zeros': 'tab:blue', 'counts': 'tab:green'}
for init, (_, _, h) in runs.items():
    ax1.plot(h['step'], h['val'], label=f'{init} init', color=colors[init])
    ax2.plot(h['step'], [v - floor_val for v in h['val']], label=f'{init} init', color=colors[init])
for ax in (ax1, ax2):
    ax.set_xlabel('step'); ax.grid(alpha=.3); ax.legend()
ax1.axhline(floor_val, color='black', ls='--', label='floor')
ax1.axhline(torch.log(torch.tensor(65.)).item(), color='gray', ls=':', label='ln 65')
ax1.set_ylabel('val loss'); ax1.set_title('Validation loss (dashed = count-based floor)')
ax1.legend()
ax2.set_yscale('log'); ax2.set_ylabel('val loss − floor (log scale)'); ax2.set_title('Gap to the floor')
plt.tight_layout(); plt.show()

print(f"{'init':>7} | {'final train':>11} | {'final val':>9} | {'gap to floor':>12}")
for init, (_, _, h) in runs.items():
    print(f"{init:>7} | {h['train'][-1]:11.4f} | {h['val'][-1]:9.4f} | {h['val'][-1] - floor_val:+12.4f}")
print(f"{'counts*':>7} | {floor_train:11.4f} | {floor_val:9.4f} | {0:+12.4f}   (* exact optimum, no training)")
```

## 5. The Q row: does P(U | Q) climb toward 0.94?

In `my_llm_1.ipynb` (lr = 1e-3, 5,000 steps, random init), `P(U | Q)` only reached 0.23.
`Q` is rare (about 230 times in train), so its row receives very few gradient updates.

```python
show = lambda c: {'\n': '\\n', ' ': "' '"}.get(c, c)
P_actual = train_counts / train_counts.sum(dim=1, keepdim=True)   # raw frequencies (no smoothing)

def compare_row(ch, top_k=5):
    i = stoi[ch]
    print(f"\n=== P(next | '{ch}')   seen {int(train_counts[i].sum())} times in train ===")
    top = torch.topk(P_actual[i], top_k).indices.tolist()
    header = f"{'next':>5} | {'actual':>6} | {'smoothed':>8}"
    for init in runs: header += f" | {init+' 0':>9} | {init+' end':>10}"
    print(header)
    for j in top:
        line = f"{show(itos[j]):>5} | {P_actual[i, j]:6.3f} | {P_counts[i, j]:8.3f}"
        for init, (m, W0, _) in runs.items():
            p0 = F.softmax(W0[i], -1)[j]
            p1 = F.softmax(m.token_embedding_table.weight.detach()[i], -1)[j]
            line += f" | {p0:9.3f} | {p1:10.3f}"
        print(line)

for ch in ['Q', 'A', 'B']:
    compare_row(ch)
```

## 6. Is the whole table right? Average gap per row vs how common the character is

```python
# KL(P_counts[i] || P_model[i]) per row: 0 means the row matches the count-based optimum exactly
row_freq = train_counts.sum(dim=1)
fig, ax = plt.subplots(figsize=(8, 4.5))
for init, (m, _, _) in runs.items():
    logQ = F.log_softmax(m.token_embedding_table.weight.detach(), -1)
    kl = (P_counts * (logP_counts - logQ)).sum(dim=1)
    ax.scatter(row_freq, kl, s=14, label=f'{init} init', color=colors[init], alpha=.75)
ax.set_xscale('log'); ax.set_yscale('log')
ax.set_xlabel('how often the character appears in train (log)')
ax.set_ylabel('KL(count optimum || model) for its row (log)')
ax.set_title('Rare characters are furthest from the optimum')
ax.legend(); ax.grid(alpha=.3); plt.tight_layout(); plt.show()
```

## 7. Result: the neural bigram converges to the count-based optimum

| model | train loss | val loss | gap to floor |
|---|---|---|---|
| count-based, +1 smoothing (**the floor**) | 2.4546 | **2.4819** | 0 |
| neural, random N(0,1) init, lr 1e-2, 20k steps | 2.4560 | 2.4890 | +0.007 |
| neural, zeros init | 2.4560 | 2.4896 | +0.008 |
| neural, counts init | 2.4560 | 2.4900 | +0.008 |
| *(my_llm_1: random init, lr 1e-3, 5k steps)* | *2.57* | *2.59* | *+0.11* |

What the numbers show:
- **All three runs land within 0.01 of the floor.** By step 5,000 they're already there. The old run's gap of 0.11 was only lack of training (lr too low, too few steps), not a limit of the model.
- **Q is fixed.** `P(U | Q)` went from 0.23 (my_llm_1) to **0.951**, matching the true frequency of 0.939. It's actually *closer* to the truth than the +1 smoothed table (0.736). Adding 1 to all 65 cells of a row that has only 230 counts flattens it a lot. The neural model learns a much milder smoothing on its own.
- **After 20,000 steps the three inits are identical** (same numbers to 3 decimals). At lr = 1e-2 with Adam, the starting point is forgotten long before the end. Init changes the start of the curve, not the end.
- **Starting at the optimum doesn't keep you there.** The `counts` run starts at val 2.4819 and drifts *up* to 2.49. Each step uses a noisy 256-example batch, and lr = 1e-2 is large enough that this noise keeps the weights jittering around the minimum instead of settling in it. That jitter is the remaining +0.007 gap. **Next step:** decay the learning rate (e.g. 1e-2 → 1e-4) at the end of training, and the gap should shrink further.

## Notes: why initialisation matters

**1. The first loss is a free sanity check.**
Before training, a model that knows nothing should predict uniformly, so the loss should be `−ln(1/vocab_size)`. For 65 characters that's 4.17.
- `zeros` gives exactly 4.17.
- PyTorch's default N(0, 1) gives about 4.7. The logits are random, so the model is *confidently wrong* about some characters.
- If the first loss is much higher than ln(V), the init is too large. If it's much lower, something is leaking (for example targets ending up in the inputs).

**2. A bad init wastes training steps.**
With random init, the first part of training only undoes the noise and pulls the logits back toward uniform. That's the "hockey stick" at the start of the loss curve. It isn't learning anything about the data. With a sensible init, every step goes into real learning.

**3. Rare tokens remember their init the longest.**
A row of the embedding table only gets a gradient when its character is the *input*. Common rows (`e`, `' '`) are updated in almost every batch and forget their starting values quickly. Rare rows (`Q`, `Z`, `X`, `&`) are updated only once in a while, so they keep looking like their init for a long time. With N(0, 1) init that means they keep random preferences (`Q → z` at step 0). With zeros init they at least start at a neutral uniform guess. You can see this in `my_llm_1.ipynb`: after 5,000 steps at lr = 1e-3, `P(U | Q)` was only 0.23. Here, with lr = 1e-2 and 20,000 steps, every row gets enough updates, so all three inits end up the same.

**4. Init sets what the network can learn in deeper models.**
In a one-layer bigram, init only costs time. In deep networks (see `makemore_part_3.ipynb`) it decides whether training works at all:
- Weights that are too large saturate `tanh` (outputs stuck at ±1, gradients ≈ 0, "dead" neurons).
- Weights that are too small make activations shrink layer after layer (vanishing signal).
- **Kaiming init** scales weights by `gain / sqrt(fan_in)` to keep activations at unit variance through every layer.
- A common trick is to make the **last layer's weights small** (e.g. × 0.01). The initial logits are then near 0, the first loss is near ln(V), and there's no hockey stick.

**5. Init from a known answer is a powerful check.**
Starting from `log P_counts` puts the model exactly at the optimum. Here, training moved it *up* slightly (2.4819 → 2.4900) instead of down. That's the clue that the remaining gap comes from **batch noise at a fixed learning rate**, not from a model that can't represent the answer. A big jump up would point to a bug instead: lr far too high, the wrong loss, or data leakage.

**6. The habit: verify models against ground truth.**
Whenever an exact or simple baseline exists (count tables, uniform guessing, a linear model), compute it first. Then check that the neural model reaches it. If it doesn't, the gap tells you where to look: rare rows, learning rate, number of steps, or regularisation such as AdamW's weight decay pulling logits toward 0.
