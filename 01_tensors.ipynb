```python
x = torch.tensor(2.0, requires_grad=True)
y = x ** 2
z = 3 * y
```

```python
print(x.grad)
```

```python
z.backward()

print(x.grad)
```

```python
x = torch.tensor(2.0, requires_grad=True)

a = x * 3
b = a ** 2
c = b + 5

c.backward()
print(x)
print(a)
print(b)
print(c)
print(x.grad)
print(c.grad_fn)
print(c.grad_fn.next_functions)
```

```python
x = torch.tensor(2.0)

a = x * 3
b = a ** 2
c = b + 5

print(x.requires_grad)
print(a.grad_fn)
print(b.grad_fn)
print(c.grad_fn)

print(x)
print(a)
print(b)
print(c)

print(c.grad_fn)
```

```python
x = torch.tensor(2.0, requires_grad=True)

a = x * 3
b = a ** 2
c = b + 5

print(x, x.grad)
# print(a, a.grad_fn, a.grad)   ---> will throw a warning
print(b)
print(c)

print(c.grad_fn)
print(c.grad_fn.next_functions)
```

```python
x = torch.tensor(2.0, requires_grad=True)

y = x ** 2

y.backward()

print(x.grad)

y.backward()

print(x.grad)
```

```python
x = torch.tensor(2.0, requires_grad=True)

y = x ** 2

y.backward()

print(x.grad)

y2 = x ** 3
y2.backward()    #gradients will get accumulated

print(x.grad)
```
