```python
import torch
```

```python
x = torch.tensor([[1.0,0],[0, 1.0], [1.0, 1.0]])
print(x.shape, x.dtype)
```

```python
x1 = torch.dot(x[1], x[0])
x2 = torch.dot(x[1], x[2])
x3 = torch.dot(x[2], x[2])
x4 = torch.dot(x[2], x[0])

print(x1.item(), x2, x3, x4)
```

```python
tra = x.T
print(tra.shape)
res = x @ tra
print(res)
```

```python
sf = res[2].exp()/res[2].exp().sum()
print(sf.sum(), sf)

k = torch.tensor([2.0, -2.0, 0.0])
sf_k = k.exp()/k.exp().sum()
print(sf_k, sf_k.sum())
```

```python
scores = x @ x.T
weights = torch.softmax(scores, dim = 1)
w_row = torch.softmax(scores, dim = 0)
print(weights)
print(" ")
print(scores)
print(" ")
print(w_row)
```

```python
out = weights @ x
print(out)
```

```python
w_q = torch.randn(2,2)
w_k = torch.randn(2,2)
w_v = torch.randn(2,2)
w_q, w_k, w_v
```

```python
Q = x @ w_q
K = x @ w_k
V = x @ w_v

print(Q,K,V, Q.shape)
```

```python
scores = Q @ K.T
print(scores)
print(" ")
attn_weights = torch.softmax(scores, dim = 1)
print(" ")
print(attn_weights)
print(" ")
print(attn_weights.sum(dim=1))
out = attn_weights @ V
print(" ")
print(out)
```

```python
for d in [2, 16, 128, 1024]:
    k = torch.randn(10000, d)
    q = torch.randn(10000, d)

    score = k * q
    summed = score.sum(dim = 1)
    stdd = torch.std(summed)
    stdd_ = d**0.5
    print(stdd.item(), stdd_)
```
